DATA VISUALIZATION -> MATPLOTLIB

```python
# LINE GRAPH
import matplotlib.pyplot as plt
import numpy as np

x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

plt.figure(figsize=(8, 5))
plt.plot(x, y, marker='o', linestyle='-', color='b', label='Prime Numbers')

plt.title('Line Graph Example')
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.legend()
plt.grid(True)

plt.show()
```

```python
# BAR GRAPH
import matplotlib.pyplot as plt

categories = ['Category A', 'Category B', 'Category C', 'Category D', 'Category E']
values = [12, 19, 7, 15, 22]

plt.figure(figsize=(8, 5))

plt.bar(categories, values, color='skyblue', edgecolor='black')

plt.title('Bar Graph Example')
plt.xlabel('Categories')
plt.ylabel('Values')

plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```

```python
# HISTOGRAM
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)
data = np.random.randn(1000)

plt.figure(figsize=(8, 5))

plt.hist(data, bins=20, color='skyblue', edgecolor='black', alpha=0.7)
plt.title('Histogram Example')
plt.xlabel('Values')
plt.ylabel('Frequency')

plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```

```python
# SCATTER PLOT
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)
x = np.random.rand(50) * 10
y = 2 * x + np.random.randn(50) * 2

plt.figure(figsize=(8, 5))

plt.scatter(x, y, color='dodgerblue', edgecolor='black', alpha=0.8, s=50, label='Data points')

plt.title('Scatter Plot Example')
plt.xlabel('X Axis')
plt.ylabel('Y Axis')

plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

print(df.head())

```

```python
numeric_cols = df.select_dtypes(include='number').columns

print("\nNumeric columns:")
print(numeric_cols.tolist())
```

```python
# 1. Correlation heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(
    df[numeric_cols].corr(),
    annot=True,
    cmap="coolwarm"
)
plt.title("Correlation Heatmap")
plt.show()
```

```python
# 2. Histogram
if len(numeric_cols) >= 1:
    sns.histplot(
        data=df,
        x=numeric_cols[0],
        kde=True
    )
    plt.title(f"Distribution of {numeric_cols[0]}")
    plt.show()
```

```python
# 3. Boxplot
plt.figure(figsize=(12, 6))
sns.boxplot(data=df[numeric_cols])
plt.xticks(rotation=45)
plt.title("Boxplot of Numeric Columns")
plt.show()
```

```python
# 5. Scatter plot
if len(numeric_cols) >= 2:
    sns.scatterplot(
        data=df,
        x=numeric_cols[0],
        y=numeric_cols[1]
    )
    plt.title(f"{numeric_cols[0]} vs {numeric_cols[1]}")
    plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

numeric_cols = df.select_dtypes(include='number').columns

print("Numeric columns:", numeric_cols.tolist())
sns.kdeplot(data=df, x=numeric_cols[0], fill=True)

plt.title(f"KDE Plot of {numeric_cols[0]}")
plt.show()
```
