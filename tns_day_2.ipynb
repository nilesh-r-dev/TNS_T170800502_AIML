```python
# Creating a list
my_list = [60, 42, 30, 40, 20,"Apple",True,1.5,"Nil"]
print("List:", my_list)
print (type(my_list))
```

```python
# append() → Adds a single element to the end of the list
numbers = [100, 20, 30]
numbers.append(25)
numbers.append(70)
print("After append:", numbers)
```

```python
# extend() → Adds all elements from another iterable (like another list)
numbers = [1, 2, 3]
print (numbers)
letter = ['A','B','C']
print(type(letter))
numbers.extend(letter)
print("After extend:", numbers)
print(type(numbers))
letter.extend(numbers)
print(letter)
```

```python
# insert() → Inserts an element at a specified index
numbers = [1, 3, 4,5]
numbers.insert(3,"abcvghavdjav")
print("After insert:", numbers)
```

```python
# remove() → Removes the first occurrence of a specified value
numbers = [1,2,5, 2,3,2,2, 3, 5,2,"apple"]
numbers.remove(5)
print("After remove:", numbers)
```

```python
# pop() → Removes and returns element at given index (default is last)
numbers = [10, 20, 30,40,50]
popped_value = numbers.pop(3)
print("Popped value:", popped_value)
print("After pop:", numbers)
```

```python
# clear() → Removes all elements from the list
numbers = [10, 20, 30]
numbers.clear()
print("After clear:", numbers)
```

```python
# index() → Returns the index of the first occurrence of a specified value
numbers = [10,10,40, 30,20 ,20, 30, 40]
index_pos = numbers.index(40)
print("Index of :", index_pos)
```

```python
# count() → Counts the number of times a value occurs in the list
numbers = [1, 2, 3, 2, 4, 2,2,2,2,2,3,3,3]
count_numbers = numbers.count(20)
print("Count of value:", count_numbers)
```

```python
# reverse() → Reverses the list in-place
numbers = [1, 2, 3, 4,"Kartik"]
numbers.reverse()
print("After reverse:", numbers)
```

```python
# sort() → Sorts the list in ascending order
numbers = [1.0,20.456,8,9,7,5,4,9]
numbers.sort()
print("After sort:", numbers)
```

```python
# copy() → Returns a shallow copy of the list
numbers = [1, 2, 3,"kartik"]
copy_numbers = numbers.copy()
print("Copied list:", copy_numbers)
```

```python
# len() → Returns the number of elements in the list
numbers = [1, 2, 3, 4,5,6,7,6]
result = len (numbers)
print("Length of list:", result)
```

```python
# sum() → Returns the sum of all numeric elements in the list
numbers = [1, 2, 3, 4,5,6.6]
total = sum(numbers)
print("Sum of list:", total)
```

```python
# max() → Returns the largest element in the list
# min() → Returns the smallest element in the list
numbers = [5, 8, 2, 10, 3,140,-9]
print("Max value:", max(numbers))
print("Min value:", min(numbers))
```

**Sets**

```python
set1 = {1, 2, 3}
set2 = {2, 3, 4}
result = set2.symmetric_difference(set1)
print("Symmetric Difference:", result)
```

```python
my_set = {10, 20, 30, 40,6,7,8}
print("Length of set:", len(my_set))
```

```python
set1 = {1, 2} #-----------Subset
set2 = {1, 2, 3,4,5}#-----Superset
print("Is set1 subset of set2?", set1.issubset(set2))
```

```python
#Intersection
set1 = {1, 2, 3,4}
set2 = {20, 33, 4,55}
result = set1.intersection(set2)
print("Intersection:", result)
```

```python
#Union
set1 = {1, 2, 3,5}
set2 = {3, 1.1,23,45,4, 5,4}
result = set2.union(set1)
print("Union:", result)
```

**Dictionaries**

```python
# Declaring a dictionary
student = {
    "name": "Sam",
    "age": "Sam",
    "ht": "Machine Learning",
     "ht": 20.0,
    "ht":123456,
}
print("Dictionary:", student)
```

```python
person = {"name": "Alex", "age": 25, "city": "Mumbai"}

# Fetch existing key
print(person.get("name"))        # Output: Alex

# Fetch non-existing key
print(person.get("country"))     # Output: None
```

```python
person = {"name": "Alex", "age": 25, "city": "Mumbai", "Country":"India"}

# Get all keys
print(person.keys())
```

```python
person = {"name": "Alex", "age": 25, "city": "Mumbai"}

# Get all values
print(person.values())
```

```python
person = {"name": "Alex", "age": 25, "city": "Mumbai"}

# Get all key-value pairs
print(person.items())
```

```python
person = {"name": "Alex", "age": 25, "city": "Mumbai"}

# Update dictionary with new data
person.update({"country":"India","age":16})
print(person)
```

```python
person = {"name": "Alex", "age": 25, "city": "Mumbai"}
print (person)

# Remove a key and get its value
removed_value = person.pop("age")
print("Removed value:", removed_value)
print(person)
```

```python
person = {"age":2345}

# Get value if exists, else set a default
value = person.setdefault("country", "India")
print(value)
print(person)
```

```python
person = {"name": "Alex", "age": 25, "city": "Mumbai"}

# Delete a specific key
del person["age"]
print(person)

# Delete the entire dictionary
del person
```

```python
person = {"name": "Alex", "age": 25, "city": "Mumbai"}

# Make a copy
person_copy = person.copy()
print("Copy:", person_copy)
```

```python
person = {"name": "Alex", "age": 25, "city": "Mumbai"}

# Clear dictionary
person.clear()
print(person)
```
