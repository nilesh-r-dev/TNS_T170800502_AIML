5)Find the factorial of a number using a loop.


```python
n = int(input("Enter the number:"))
factorial = 1

if n < 0:
  print("Factorial does not exist for negative numbers")
elif n == 0:
  print("The factorial of 0 is 1")
else:
  for i in range(1, n + 1):
    factorial = factorial * i
  print("The factorial of", n, "is", factorial)
```

7)Reverse a number using a loop.


```python
n = int(input("Enter your number:"))
original_num = n
reversed_num = 0

while n > 0:
  digit = n % 10
  reversed_num = reversed_num * 10 + digit
  n = n // 10

print("The reverse of", original_num, "is", reversed_num)
```

8)Check whether a number is prime.


```python
n= int(input("Enter your number:"))
if n > 1:
  for i in range(2, int(n/2)+1):
    if (n % i) == 0:
      print(n, "is not a prime number")
      break
  else:
    print(n, "is a prime number")
```

9)Write a function to check if a number is palindrome.


```python
n = int(input("Enter your number:"))
original_num = n
reversed_num = 0

while n > 0:
  digit = n % 10
  reversed_num = reversed_num * 10 + digit
  n = n // 10
if original_num == reversed_num:
  print(original_num, "is a palindrome number")
else:
  print(original_num, "is not a palindrome number")
```

10)Write a function to calculate sum of digits of a number.


```python
n = int(input("Enter your number:"))
original_num = n
digit_sum = 0

abs_n = abs(n)

while abs_n > 0:
  digit = abs_n % 10
  digit_sum += digit
  abs_n //= 10

print("The sum of digits of", original_num, "is", digit_sum)
```

23)Count frequency of characters in a string.


```python
text = input("Enter a string: ")

frequency = {}

for char in text:
    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1

print(frequency)
```

25)Sort dictionary by values.


```python
my_dict = {"apple": 3, "banana": 1, "cherry": 2, "date": 4}

sorted_items_asc = sorted(my_dict.items(), key=lambda item: item[1])
sorted_dict_asc = dict(sorted_items_asc)

print("Original dictionary:", my_dict)
print("Sorted by values (ascending):", sorted_dict_asc)

sorted_items_desc = sorted(my_dict.items(), key=lambda item: item[1], reverse=True)
sorted_dict_desc = dict(sorted_items_desc)

print("Sorted by values (descending):", sorted_dict_desc)
```

27)Create a class Rectangle to calculate area and perimeter.


```python
class rectangle:
  def __init__(self, length, width):
    self.length = length
    self.width = width
  def calculate_area(self):
    return self.length * self.width
  def calculate_perimeter(self):
    return 2 * (self.length + self.width)
rectangle1 = rectangle(5, 3)
area = rectangle1.calculate_area()
perimeter = rectangle1.calculate_perimeter()
print("Area:", area)
print("Perimeter:", perimeter)
```

29)Create a class ShoppingCart using dictionary (item: price) and calculate total bill.

```python
class ShoppingCart:
  def __init__(self):
    self.items = {}

  def add_item(self, item_name, price):
    self.items[item_name] = price

  def calculate_total(self):
    total_bill = 0
    for price in self.items.values():
      total_bill += price
    return total_bill

my_cart = ShoppingCart()
my_cart.add_item("Laptop", 1200)
my_cart.add_item("Mouse", 25)
my_cart.add_item("Keyboard", 75)

print("Shopping Cart Items:", my_cart.items)
print("Total Bill:", my_cart.calculate_total())
```

30)Create a class Library to store books in a list and add/remove books.


```python
class Library:
  def __init__(self):
    self.books = []

  def add_book(self, book_name):
    self.books.append(book_name)
    print(f"'{book_name}' has been added to the library.")

  def remove_book(self, book_name):
    if book_name in self.books:
      self.books.remove(book_name)
      print(f"'{book_name}' has been removed from the library.")
    else:
      print(f"'{book_name}' is not in the library.")

  def display_books(self):
    if not self.books:
      print("The library is empty.")
    else:
      print("Books in the library:")
      for book in self.books:
        print(f"- {book}")

my_library = Library()
my_library.add_book("The Great Gatsby")
my_library.add_book("1984")
my_library.add_book("To Kill a Mockingbird")
my_library.display_books()

my_library.remove_book("1984")
my_library.display_books()

my_library.remove_book("Moby Dick")
my_library.display_books()
```
