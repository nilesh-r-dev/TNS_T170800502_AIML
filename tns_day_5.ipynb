```python
# Fill missing numerical values
for col in df.select_dtypes(include=np.number).columns:
    df[col] = df[col].fillna(df[col].median())

# Fill missing categorical values
for col in df.select_dtypes(include=['object', 'category']).columns:
    if df[col].isnull().any() and not df[col].mode().empty:
        df[col] = df[col].fillna(df[col].mode()[0])
```

```python
# Count duplicate rows
print("Duplicates:", df.duplicated().sum())

# Remove duplicate rows
df = df.drop_duplicates()

# Check updated shape
print(df.shape)
```

```python
# Remove extra spaces and convert text to lowercase
for col in df.select_dtypes(include=['object', 'category']).columns:
    df[col] = df[col].apply(
        lambda x: x.strip().lower() if isinstance(x, str) else x
    )

# Display cleaned data
print(df.head())
```

```python
# Check unique values in categorical columns
for col in df.select_dtypes(include=['object', 'category']).columns:
    print(col, ":", df[col].unique())

# Check numerical statistics
print(df.describe())
```

```python
# Convert categorical columns into numerical columns
df_encoded = pd.get_dummies(
    df,
    columns=df.select_dtypes(include=['object', 'category']).columns,
    dtype=int
)

# Display encoded data
print(df_encoded.head())

# Check final shape
print(df_encoded.shape)
```

```python
# Final dataset information
print("Dataset shape:", df_encoded.shape)

# Check missing values
print("Missing values:", df_encoded.isnull().sum().sum())

# Check duplicate rows
print("Duplicate rows:", df_encoded.duplicated().sum())

# Display final data
print(df_encoded.head())
```
